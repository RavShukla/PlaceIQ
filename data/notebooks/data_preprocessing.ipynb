```python
import numpy as np 
import pandas as pd
```

```python
df=pd.read_csv("C:/Users/gaura/Desktop/PlaceIQ/data/raw/placementdata.csv")
```

```python
#we will preprocess first column having objects ExtracurricularActivities
df["ExtracurricularActivities"].value_counts()
```

```python
mapping = {"No" : 0,
           "Yes" : 1
          }
df["ExtracurricularActivities"]= df["ExtracurricularActivities"].map(mapping)
df["ExtracurricularActivities"]
```

```python
#we will preprocess first column having objects PlacementTraining
df["PlacementTraining"].value_counts()
```

```python
mapping = {"No" : 0,
           "Yes" : 1
          }
df["PlacementTraining"]= df["PlacementTraining"].map(mapping)
df["PlacementTraining"]
```

```python
#we will preprocess first column having objects PlacementStatus
df["PlacementStatus"].value_counts()
```

```python
mapping = {"NotPlaced" : 0,
           "Placed" : 1
          }
df["PlacementStatus"]= df["PlacementStatus"].map(mapping)
df["PlacementStatus"]
```
